### Распад разрыва с уравнением состояния идеального газа
Для простоты будем рассмматривать одномерной задачу в неподвижной декратовой системе координат.

Пусть справа от разрыва давление в среде $P_0$, массовая скорость $U_0$, а слева - давление $P_1$, массовая скорость $U_1$.
Предположим, что сам разрыв перемещается по пространству в лабораторной системе координат со скоростью $D_0$.
В этом случае законы сохранения массы и импульса на разрыве могут быть записаны в следующем виде:
$$\rho_1 (U_1 - D_0) = \rho_0 (U_0 - D_0)$$
$$P_1 + \rho_1 (U_1 - D_0)^2 = P_0 + \rho_0 (U_0 - D_0)^2$$
где $\rho_0$ и $\rho_1$ - это значение плотности справа и слева от повержности разрыва.

Выражая $(U_1 - D_0)$ из первого уравнения и подставляя во второе уравнение и разрешая относительно $(U_0 - D_0)^2$, получаем:

```python
from sympy import symbols, solve, Symbol

# Symbols
rho_0, rho_1, P_0, P_1, U_0, U_1, D_0, D_1 = symbols("rho_0 rho_1 P_0 P_1 U_0 U_1 D_0 D_1")

# Expressions
eq1 = rho_1*(U_1 - D_0) - rho_0*(U_0-D_0)
eq2 = P_1 + rho_1*(U_1 - D_0)**2 - P_0 - rho_0*(U_0 - D_0)**2

# Substitution and elimination
z = Symbol('z')
eq1.subs((U_1 - D_0), z)
eq2.subs((U_1 - D_0), z)
```
